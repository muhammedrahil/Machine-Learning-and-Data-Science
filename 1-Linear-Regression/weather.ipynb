```python
import pandas as pd

dataset = pd.read_csv('large_min_max_temp_dataset.csv')
```

```python
print(dataset.shape)
```

```python
print(dataset.describe())
```

```python
dataset = dataset.rename(columns={
    'Min Temp (°C)': 'min_temp',
    'Max Temp (°C)': 'max_temp',
})
```

```python
dataset.plot(x='min_temp',y='max_temp',style='o')
```

```python
import matplotlib.pyplot as plt

plt.title("Min vs Max Temperature")
```

```python
plt.xlabel('Min Temperature')
plt.ylabel('Max Temperature')
```

```python
plt.show()
```

```python
plt.figure(figsize=(10,10))
plt.tight_layout()
```

```python
import seaborn as seaborninstance

seaborninstance.displot(x=dataset['min_temp'])
```

```python
x = dataset['min_temp'].values.reshape(-1,1)
y = dataset['max_temp'].values.reshape(-1,1)
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.2, random_state=0)
```

```python

from sklearn.linear_model import LinearRegression
regression = LinearRegression()
regression.fit(x_train,y_train)

```

```python
print(regression.intercept_)
```

```python
print(regression.coef_)
```

```python
y_pred = regression.predict(x_test)
```
